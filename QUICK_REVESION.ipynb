```python

# ============================================================
# MACHINE LEARNING — COMPLETE PRACTICAL NOTES
# Focus: Algorithms and concepts used in real-world projects
# ============================================================


# ============================================================
# 1. MACHINE LEARNING FUNDAMENTALS
# ============================================================

# Machine Learning (ML):
# Enables computers to learn patterns from data and make predictions.

# Supervised Learning:
# Learns from labeled data (X, y).
# Uses: Price prediction, churn prediction, fraud detection.

# Unsupervised Learning:
# Finds patterns in unlabeled data.
# Uses: Customer segmentation, anomaly exploration, compression.

# Reinforcement Learning:
# Learns actions using rewards and penalties.
# Uses: Robotics, control systems, simulations.

# Self-Supervised Learning:
# Generates learning signals from the data itself.
# Uses: Language models, computer vision, representation learning.

# Transfer Learning:
# Reuses a pretrained model for a related task.
# Uses: Image classification, NLP, pretrained neural networks.

# Online Learning:
# Updates a model incrementally as new data arrives.
# Uses: Streaming data and changing environments.

# Semi-Supervised Learning:
# Uses a small labeled dataset with a larger unlabeled dataset.

# Important terms:
# Feature (X): Input variable.
# Target (y): Value to predict.
# Training: Learning patterns from data.
# Inference: Predicting on new data.
# Parameters: Values learned by a model.
# Hyperparameters: Settings chosen before/during training.
# Loss: Measures prediction error during training.
# Generalization: Performance on unseen data.
# Overfitting: Learns noise or training-specific patterns.
# Underfitting: Fails to learn important patterns.
# Baseline: Simple model used for comparison.
# Data Leakage: Unavailable or forbidden information enters training.


# ============================================================
# 2. REAL-WORLD MACHINE LEARNING WORKFLOW
# ============================================================

# 1. Define the business problem.
# 2. Collect data.
# 3. Validate and clean data.
# 4. Perform exploratory data analysis (EDA).
# 5. Define features and target.
# 6. Split data into training, validation and test sets.
# 7. Build preprocessing pipelines.
# 8. Train a baseline model.
# 9. Train and compare candidate models.
# 10. Evaluate using suitable metrics.
# 11. Tune hyperparameters.
# 12. Test the selected model on unseen data.
# 13. Save and deploy the model.
# 14. Monitor predictions and data drift.
# 15. Retrain when justified by new data or degradation.


# ============================================================
# 3. DATA PREPROCESSING
# ============================================================

# Missing Values:
# Numerical data -> Median or mean imputation.
# Categorical data -> Most frequent value or missing category.
# Drop data only when justified.

# Numerical Encoding:
# OneHotEncoder -> Nominal categories without natural order.
# OrdinalEncoder -> Categories with a meaningful order.
# Target Encoding -> High-cardinality categories; prevent leakage.

# Feature Scaling:
# StandardScaler -> Standardizes features.
# MinMaxScaler -> Scales values to a fixed range.
# RobustScaler -> Uses robust statistics; helpful with outliers.
# Tree models generally do not require feature scaling.

# Outliers:
# IQR -> Detects values outside the interquartile range.
# Z-score -> Measures distance from the mean in standard deviations.
# Do not automatically remove valid extreme observations.

# Feature Engineering:
# Extract date parts, calculate ratios, create counts and aggregates.
# Create domain-specific features that improve predictions.

# Feature Selection:
# Remove irrelevant identifiers, constant features and duplicates.
# Use domain knowledge, regularization or permutation importance.

# Leakage Prevention:
# Split data before fitting preprocessing.
# Fit imputers, scalers and encoders on training data only.
# Use chronological splits for future prediction tasks.


# ============================================================
# 4. REGRESSION ALGORITHMS
# ============================================================

# Regression predicts numerical values.

# 1. Linear Regression
# Models a linear relationship between features and target.
# Use: House prices, sales forecasting baselines.

# 2. Ridge Regression (L2)
# Adds a squared-coefficient penalty.
# Use: Reduce overfitting and stabilize correlated features.

# 3. Lasso Regression (L1)
# Can shrink some coefficients to zero.
# Use: Regularization and feature selection.

# 4. Elastic Net
# Combines L1 and L2 regularization.
# Use: High-dimensional data with correlated features.

# 5. Decision Tree Regressor
# Learns nonlinear rules by splitting features.
# Use: Nonlinear tabular prediction; interpretable rules.

# 6. Random Forest Regressor
# Averages predictions from many randomized trees.
# Use: Robust nonlinear predictions on tabular data.

# 7. Gradient Boosting Regressor
# Builds trees sequentially to improve earlier predictions.
# Use: Structured/tabular prediction.

# 8. XGBoost / LightGBM / CatBoost
# Efficient gradient-boosted tree libraries.
# Use: Strong tabular baselines and competitive production models.
# CatBoost is particularly convenient with categorical features.

# 9. Support Vector Regression (SVR)
# Fits a function within a tolerance around observed values.
# Use: Smaller or medium datasets with suitable feature scaling.

# 10. K-Nearest Neighbors Regressor (KNN)
# Predicts from nearby training examples.
# Use: Small datasets with meaningful distance relationships.

# 11. Neural Network Regressor
# Learns complex nonlinear relationships.
# Use: Large or complex datasets when justified by performance.


# ============================================================
# 5. CLASSIFICATION ALGORITHMS
# ============================================================

# Classification predicts classes or class probabilities.

# 1. Logistic Regression
# Estimates class probabilities.
# Use: Churn, risk prediction, interpretable classification.

# 2. Decision Tree Classifier
# Learns rule-based splits.
# Use: Nonlinear classification and explainable rules.

# 3. Random Forest Classifier
# Combines predictions from many randomized trees.
# Use: General-purpose tabular classification.

# 4. Gradient Boosting Classifier
# Sequentially builds models to correct prediction errors.
# Use: Structured classification tasks.

# 5. XGBoost / LightGBM / CatBoost
# Optimized gradient-boosted tree implementations.
# Use: Tabular classification, ranking and risk scoring.

# 6. Support Vector Machine (SVM)
# Finds a decision boundary with a large margin.
# Use: Smaller or medium datasets; scale features appropriately.

# 7. K-Nearest Neighbors (KNN)
# Predicts using nearby examples.
# Use: Small datasets where distance is meaningful.

# 8. Naive Bayes
# Uses Bayes' theorem with conditional independence assumptions.
# Use: Text classification, spam filtering, fast baselines.

# 9. Neural Network Classifier
# Learns nonlinear decision boundaries.
# Use: Complex or large datasets and learned representations.

# 10. Linear Discriminant Analysis (LDA)
# Models class distributions and separates classes.
# Use: Certain smaller, approximately Gaussian datasets.

# 11. Calibrated Classification
# Improves agreement between predicted probabilities and observed
# event frequencies.
# Use: Risk scores and probability-based decisions.


# ============================================================
# 6. UNSUPERVISED LEARNING
# ============================================================

# Clustering:
# Groups observations without target labels.

# 1. K-Means
# Groups data around cluster centroids.
# Use: Customer segmentation; scale features when appropriate.

# 2. DBSCAN
# Finds dense groups and labels isolated points as noise.
# Use: Irregular-shaped clusters and spatial data.

# 3. Hierarchical Clustering
# Builds a hierarchy of clusters.
# Use: Exploring relationships and small-to-medium datasets.

# Dimensionality Reduction:
# Reduces the number of features while preserving useful structure.

# 4. PCA (Principal Component Analysis)
# Creates orthogonal directions capturing maximum variance.
# Use: Compression, visualization, reducing correlated dimensions.

# 5. Truncated SVD
# Reduces sparse matrices without requiring dense conversion.
# Use: Text features and sparse high-dimensional data.

# 6. UMAP / t-SNE
# Nonlinear techniques for low-dimensional visualization.
# Use: Exploring complex high-dimensional structure.
# Do not automatically treat visual clusters as proven groups.

# Anomaly Detection:
# Finds observations that differ substantially from normal behavior.

# 7. Isolation Forest
# Isolates unusual observations through randomized trees.
# Use: Fraud investigation, operational anomalies.

# 8. Local Outlier Factor (LOF)
# Compares local density with neighboring observations.
# Use: Local anomalies in suitable datasets.


# ============================================================
# 7. ENSEMBLE LEARNING
# ============================================================

# Ensemble:
# Combines multiple models to improve performance or robustness.

# Bagging:
# Trains models on randomized samples and combines predictions.
# Example: Random Forest.

# Boosting:
# Builds models sequentially to correct earlier errors.
# Examples: AdaBoost, Gradient Boosting, XGBoost, LightGBM, CatBoost.

# Voting:
# Combines predictions from different models.
# Classification: Hard or soft voting.

# Stacking:
# Uses predictions from base models as inputs to a meta-model.
# Use: Combine complementary models; use out-of-fold predictions
# to prevent leakage.

# Practical priority:
# Random Forest and gradient-boosted trees are strong tabular choices.
# Compare them against a simple baseline instead of assuming
# an ensemble will always perform better.


# ============================================================
# 8. MODEL EVALUATION — REGRESSION
# ============================================================

# MAE (Mean Absolute Error):
# Average absolute prediction error.
# Easy to interpret in target units.

# MSE (Mean Squared Error):
# Squares errors, penalizing large errors more heavily.

# RMSE (Root Mean Squared Error):
# Square root of MSE; expressed in target units.

# R-squared (R²):
# Measures improvement relative to predicting the target mean.
# Can be negative on test data.

# MAPE:
# Percentage-based error; problematic when actual values are zero
# or close to zero.

# Always compare metrics against a baseline and business requirements.


# ============================================================
# 9. MODEL EVALUATION — CLASSIFICATION
# ============================================================

# Confusion Matrix:
# True Positive (TP): Correct positive prediction.
# True Negative (TN): Correct negative prediction.
# False Positive (FP): Incorrect positive prediction.
# False Negative (FN): Missed positive case.

# Accuracy:
# Correct predictions / Total predictions.
# Misleading when classes are highly imbalanced.

# Precision:
# TP / (TP + FP)
# Important when false alarms are costly.

# Recall (Sensitivity):
# TP / (TP + FN)
# Important when missing positive cases is costly.

# F1 Score:
# Harmonic mean of precision and recall.

# Specificity:
# TN / (TN + FP)

# ROC-AUC:
# Measures ranking ability across classification thresholds.

# PR-AUC / Average Precision:
# Particularly informative for rare positive classes.

# Log Loss:
# Evaluates predicted probabilities and penalizes confident errors.

# Threshold Tuning:
# Choose a decision threshold based on business costs and constraints.
# Do not assume 0.5 is always optimal.

# Probability Calibration:
# Check whether predicted probabilities match actual frequencies.


# ============================================================
# 10. DATA SPLITTING AND VALIDATION
# ============================================================

# Training Set:
# Used to fit model parameters.

# Validation Set:
# Used for model selection and hyperparameter tuning.

# Test Set:
# Used for final evaluation after decisions are complete.

# Train/Test Split:
# A simple random split for independent, identically distributed data.

# Stratified Split:
# Preserves class proportions in classification.

# K-Fold Cross-Validation:
# Trains and validates across multiple data folds.

# Stratified K-Fold:
# Preserves approximate class proportions in each fold.

# Group Cross-Validation:
# Keeps related observations together.
# Use: Multiple records from the same person, patient or customer.

# Time-Series Validation:
# Preserves time order; train on past and validate on future.

# Nested Cross-Validation:
# Separates hyperparameter selection from performance estimation.
# Useful when an unbiased comparison is particularly important.

# Never repeatedly tune against the final test set.


# ============================================================
# 11. HYPERPARAMETER TUNING
# ============================================================

# GridSearchCV:
# Exhaustively evaluates a specified parameter grid.
# Use: Small, manageable search spaces.

# RandomizedSearchCV:
# Samples parameter combinations.
# Use: Larger search spaces with limited computation.

# Bayesian Optimization:
# Selects promising configurations using previous trial results.
# Use: Expensive model training or larger tuning tasks.

# Optuna:
# Practical library for hyperparameter optimization.

# Early Stopping:
# Stops training when validation performance stops improving.
# Use: Boosting and neural networks where supported.

# Tune only on training/validation data or within cross-validation.


# ============================================================
# 12. FEATURE ANALYSIS AND STATISTICS
# ============================================================

# Correlation:
# Measures association; does not establish causation.

# Pearson Correlation:
# Measures linear association between numerical variables.

# Spearman Correlation:
# Measures rank-based monotonic association.

# Multicollinearity:
# Strong relationships among input features.
# Can make linear-model coefficients unstable.

# VIF:
# Diagnostic for multicollinearity in regression features.

# Statistical Tests:
# t-test -> Compare means under appropriate assumptions.
# Mann-Whitney U -> Compare distributions/ranks for independent groups.
# ANOVA -> Compare means across multiple groups.
# Chi-Square -> Test association between categorical variables.
# Shapiro-Wilk -> Test evidence against normality.
# Levene's Test -> Assess equality of variances.
# These tests support analysis; they are not ML algorithms.

# Permutation Importance:
# Measures performance change after shuffling a feature.

# Partial Dependence / SHAP:
# Tools for understanding model behavior.
# Interpret carefully when features are correlated.


# ============================================================
# 13. SCIKIT-LEARN PRODUCTION PIPELINES
# ============================================================

# Pipeline:
# Combines preprocessing and model training into one workflow.

# ColumnTransformer:
# Applies different preprocessing to different feature columns.

# Benefits:
# - Reduces inconsistent preprocessing.
# - Helps prevent leakage during cross-validation.
# - Makes training and inference more reproducible.
# - Simplifies saving and deploying complete workflows.

# Example:
#
# from sklearn.compose import ColumnTransformer
# from sklearn.pipeline import Pipeline
# from sklearn.impute import SimpleImputer
# from sklearn.preprocessing import StandardScaler, OneHotEncoder
# from sklearn.linear_model import LogisticRegression
#
# numeric_pipeline = Pipeline([
#     ("imputer", SimpleImputer(strategy="median")),
#     ("scaler", StandardScaler())
# ])
#
# categorical_pipeline = Pipeline([
#     ("imputer", SimpleImputer(strategy="most_frequent")),
#     ("encoder", OneHotEncoder(handle_unknown="ignore"))
# ])
#
# preprocessor = ColumnTransformer([
#     ("num", numeric_pipeline, numerical_columns),
#     ("cat", categorical_pipeline, categorical_columns)
# ])
#
# model = Pipeline([
#     ("preprocessing", preprocessor),
#     ("classifier", LogisticRegression(max_iter=1000))
# ])
#
# model.fit(X_train, y_train)
# predictions = model.predict(X_test)


# ============================================================
# 14. CLASS IMBALANCE
# ============================================================

# Problem:
# A rare class may receive poor predictions despite high accuracy.

# Techniques:
# - Use stratified splitting.
# - Evaluate precision, recall, F1 and PR-AUC.
# - Adjust the decision threshold.
# - Use class_weight="balanced" when supported.
# - Try oversampling or undersampling when appropriate.
# - SMOTE generates synthetic minority examples for suitable data.

# Important:
# Apply resampling only inside the training process.
# Never resample the test set.
# Optimize for the real business objective, not accuracy alone.


# ============================================================
# 15. TIME SERIES AND FORECASTING
# ============================================================

# Time-series data:
# Observations recorded over time.

# Important concepts:
# Trend -> Long-term direction.
# Seasonality -> Repeating patterns.
# Lag Features -> Values from previous time steps.
# Rolling Features -> Statistics over past time windows.
# Data Leakage -> Future information accidentally enters features.

# Practical methods:
# - Naive / seasonal-naive baseline.
# - Linear regression with lag and calendar features.
# - Random Forest or gradient boosting with lag features.
# - ARIMA / SARIMA for suitable statistical forecasting problems.
# - Exponential smoothing for suitable trend/seasonal patterns.
# - Specialized neural forecasting models for justified use cases.

# Evaluation:
# Use chronological validation.
# MAE and RMSE are common metrics.
# MAPE is unsafe when actual values can be zero.
# Compare against a simple forecasting baseline.


# ============================================================
# 16. DEEP LEARNING — IMPORTANT FOUNDATIONS
# ============================================================

# Deep Learning:
# Uses neural networks with multiple layers.

# Perceptron / Dense Layer:
# Combines inputs using learned weights and biases.

# Activation Functions:
# ReLU -> Common hidden-layer activation.
# Sigmoid -> Often used for binary output probabilities.
# Softmax -> Converts multiclass logits into probabilities.

# Loss Functions:
# MSE -> Common regression loss.
# Binary Cross-Entropy -> Binary classification.
# Cross-Entropy -> Multiclass classification.

# Backpropagation:
# Computes gradients of loss with respect to model parameters.

# Gradient Descent:
# Updates parameters to reduce the loss.

# Optimizers:
# SGD -> Stochastic Gradient Descent.
# Adam -> Adaptive optimizer commonly used in deep learning.

# Epoch:
# One pass through the training dataset.

# Batch:
# A subset of data processed in one training step.

# Regularization:
# Dropout, weight decay and early stopping help control overfitting.

# Important architectures:
# MLP -> General neural networks for structured inputs.
# CNN -> Images and spatial patterns.
# RNN / LSTM / GRU -> Sequential modeling; useful in selected tasks.
# Transformers -> Language, vision and other sequence-related tasks.

# Practical rule:
# For ordinary tabular data, start with gradient-boosted trees.
# Use deep learning when data, task complexity or pretrained models
# provide a compelling reason.


# ============================================================
# 17. NLP, COMPUTER VISION AND MODERN AI
# ============================================================

# NLP (Natural Language Processing):
# Text classification, sentiment analysis, entity extraction.

# TF-IDF:
# Represents text using weighted word importance.
# Useful baseline for text classification.

# Embeddings:
# Dense vectors representing semantic or learned relationships.

# Transformers:
# Architecture based on attention.
# Foundation for many modern language models.

# Transfer Learning:
# Fine-tune or adapt a pretrained model instead of training from zero.

# Computer Vision:
# Image classification, object detection and image segmentation.
# CNNs and pretrained vision transformers are common approaches.

# Generative AI:
# Models generate text, images, audio or other content.
# Includes language models and diffusion models.

# Retrieval-Augmented Generation (RAG):
# Retrieves relevant external information before generating an answer.

# These areas are important, but are specialized extensions of
# classical ML rather than mandatory algorithms for every project.


# ============================================================
# 18. MODEL DEPLOYMENT AND MLOPS
# ============================================================

# Model Serialization:
# Save a trained model and preprocessing pipeline.
# Common Python tools: joblib and pickle.
# Load only trusted model files.

# Inference API:
# Serve predictions through an API.
# Common tools: FastAPI, Flask.

# Frontend:
# Provides a user interface for submitting inputs and viewing results.
# Example: Streamlit.

# Containers:
# Docker packages an application and its dependencies.

# Deployment:
# Host the API and frontend on suitable cloud infrastructure.

# Versioning:
# Track code, data, model artifacts and configuration.

# Experiment Tracking:
# Record parameters, metrics and model versions.
# Examples: MLflow and Weights & Biases.

# Data Drift:
# Input data distribution changes over time.

# Concept Drift:
# Relationship between inputs and targets changes over time.

# Monitoring:
# Track errors, latency, data quality, prediction distributions,
# and performance when ground-truth labels become available.

# Retraining:
# Retrain when justified by monitoring, new data and validation.


# ============================================================
# 19. IMPORTANT ENGINEERING PRACTICES
# ============================================================

# Reproducibility:
# Fix random seeds where appropriate and record configurations.

# Data Validation:
# Check data types, ranges, missing values and schema changes.

# Modular Code:
# Separate data loading, preprocessing, training, evaluation
# and inference into reusable modules.

# Testing:
# Test preprocessing, prediction shapes, API inputs and edge cases.

# Reproducible Environments:
# Pin dependencies and use a requirements file.

# Model Registry:
# Track approved model versions and deployment status.

# Security:
# Validate user inputs and avoid exposing secrets.

# Explainability:
# Understand how predictions influence business decisions.

# Human Oversight:
# Keep appropriate review and escalation for high-impact decisions.


# ============================================================
# 20. IMPORTANT LIBRARIES
# ============================================================

# NumPy          -> Numerical computing.
# Pandas         -> Data manipulation.
# Matplotlib     -> Visualization.
# Seaborn        -> Statistical visualization.
# SciPy          -> Scientific computing and statistical tests.
# Statsmodels    -> Statistical modeling and inference.
# Scikit-learn   -> Preprocessing, classical ML, evaluation, pipelines.
# XGBoost        -> Gradient-boosted trees.
# LightGBM       -> Efficient gradient-boosted trees.
# CatBoost       -> Gradient boosting with categorical-feature support.
# Imbalanced-learn -> Resampling and imbalanced classification.
# Optuna         -> Hyperparameter optimization.
# SHAP           -> Model explanations.
# MLflow         -> Experiment tracking and model lifecycle.
# PyTorch        -> Deep learning.
# FastAPI        -> Model-serving APIs.
# Streamlit      -> Interactive ML applications.
# Docker         -> Containerization.


# ============================================================
# 21. WHAT TO LEARN FIRST — PRACTICAL PRIORITY
# ============================================================

# LEVEL 1 — MUST KNOW
# - Train/test split and cross-validation.
# - Data cleaning, EDA and feature engineering.
# - Missing values, encoding and scaling.
# - Linear Regression, Ridge, Lasso and Logistic Regression.
# - Decision Trees and Random Forest.
# - Gradient Boosting, XGBoost, LightGBM and CatBoost.
# - Classification and regression metrics.
# - Leakage prevention, pipelines and hyperparameter tuning.
# - Model interpretation and error analysis.

# LEVEL 2 — IMPORTANT SPECIALIZATIONS
# - PCA and clustering (K-Means, DBSCAN).
# - Isolation Forest and anomaly detection.
# - Probability calibration and class imbalance.
# - Time-series forecasting and temporal validation.
# - Experiment tracking, deployment and model monitoring.

# LEVEL 3 — WHEN YOUR PROJECT REQUIRES THEM
# - SVM, KNN, Naive Bayes and LDA.
# - Advanced dimensionality reduction.
# - Deep learning with PyTorch.
# - NLP, computer vision and transformers.
# - Reinforcement learning.
# - Distributed training and advanced MLOps.

# ============================================================
# FINAL PRINCIPLE
# ============================================================

# There is no universally best algorithm.
# Start with a baseline, select suitable metrics, validate correctly,
# compare realistic candidates, analyze errors and deploy only after
# confirming that the model solves the actual problem.
# ============================================================
```
