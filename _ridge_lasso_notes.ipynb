```python
# RIDGE & LASSO — NOTES

# REGULARIZATION
# - Used to reduce overfitting by penalizing large coefficients.
# - It adds a penalty to the Linear Regression loss function.

# --------------------------------------------------

# L1 PENALTY → LASSO

# Penalty:
#     alpha × Σ|coefficient|

# - Uses the absolute value of coefficients.
# - Can shrink coefficients exactly to 0.
# - Therefore, Lasso can perform FEATURE SELECTION.

# MINI PROJECT:
# - We created 10 features: X1 → X10.
# - Only X1 → X5 were actually used to create the target y.
# - X6 → X10 were irrelevant/random features.

# Example:
#     X6 → small coefficient
#     X7 → small coefficient
#     X8 → small coefficient

# Lasso can turn them into:

#     X6 → 0
#     X7 → 0
#     X8 → 0

# Meaning:
#     "These features are not contributing enough,
#      so remove their effect."

# --------------------------------------------------

# L2 PENALTY → RIDGE

# Penalty:
#     alpha × Σ(coefficient²)

# - Squares the coefficients.
# - Shrinks coefficients toward 0.
# - Usually does NOT make them exactly 0.
# - Keeps all features but reduces their influence.
# - Especially useful when features are correlated.

# MINI PROJECT:

# We deliberately created correlated features:

#     X1 ↔ X2
#     X3 ↔ X4

# Ridge handles these correlated features by
# shrinking their coefficients rather than removing them.

# Example:

#     X1 → 44.1
#     X2 → 30.6
#     X3 → 59.5
#     X4 → 31.1

# While irrelevant features had coefficients close to 0:

#     X6 → -0.02
#     X7 → -0.17
#     X8 → -0.10

# They are small, but Ridge generally keeps them
# instead of forcing them to exactly 0.

# --------------------------------------------------

# ALPHA

# - Controls the strength of regularization.

#     Higher alpha
#         ↓
#     Stronger penalty
#         ↓
#     More coefficient shrinking

#     Lower alpha
#         ↓
#     Weaker penalty
#         ↓
#     Closer to Linear Regression

# Example:

#     Ridge(alpha=1)
#     Ridge(alpha=100)

# alpha=100 applies a much stronger penalty.

# --------------------------------------------------

# OUR  Implementation PIPELINE

# 1. Created dataset using NumPy
# 2. Separated X (features) and y (target)
# 3. Train/Test Split
# 4. StandardScaler
# 5. Linear Regression → baseline
# 6. Ridge → L2 regularization
# 7. Lasso → L1 regularization
# 8. Compared coefficients
# 9. Evaluated using:
#        MSE
#        RMSE
#        R²

# --------------------------------------------------

# MAIN DIFFERENCE

# RIDGE → L2
#     Shrinks coefficients
#     ↓
#     Usually keeps all features

# LASSO → L1
#     Shrinks coefficients
#     ↓
#     Can make coefficients = 0
#     ↓
#     Feature Selection

# --------------------------------------------------

# IMPORTANT

# - Always scale features before Ridge/Lasso.
# - Ridge is useful for correlated features.
# - Lasso is useful when feature selection is desired.
# - alpha controls regularization strength.
# - Ridge and Lasso are both extensions of Linear Regression.
```

```python
# RIDGE & LASSO — MATHEMATICAL INTUITION + WHAT EACH TERM MEANS

# --------------------------------------------------
# FIRST: UNDERSTAND THE LINEAR REGRESSION EQUATION
# --------------------------------------------------

#     ŷ = β₀ + β₁X₁ + β₂X₂ + ... + βₙXₙ

# Where:

#     ŷ       = predicted target value

#     β₀      = intercept
#              → predicted y when all X values are 0

#     β₁, β₂  = COEFFICIENTS
#     ...
#     βₙ      = COEFFICIENTS

#     Coefficient = SLOPE / effect of a feature
#                   on the prediction

#     X₁, X₂... = input FEATURES

# Example from our mini-project:

#     X1 → feature
#     X2 → feature
#     ...
#     X10 → feature

#     y → TARGET

# If:

#     β₁ = 44

# then X1 has a coefficient of 44.

# Because we STANDARDIZED our features, this means:

#     A 1-standard-deviation increase in X1
#     is associated with about a 44-unit increase
#     in the predicted y, keeping other features fixed.

# --------------------------------------------------
# LOSS / ERROR
# --------------------------------------------------

# Linear Regression tries to minimize:

#     Σ(yᵢ - ŷᵢ)²

# What each term means:

#     yᵢ  = actual target value for observation i

#     ŷᵢ  = predicted target value for observation i

#     yᵢ - ŷᵢ
#          = prediction ERROR / RESIDUAL

#     (...)²
#          = squares the error

#     Σ
#          = SUM all observations' squared errors

# So:

#     Σ(yᵢ - ŷᵢ)²

# means:

#     "Add up all the squared prediction errors."

# --------------------------------------------------
# RIDGE — L2 REGULARIZATION
# --------------------------------------------------

# Ridge minimizes:

#     Σ(yᵢ - ŷᵢ)² + α Σβⱼ²


# FIRST PART:

#     Σ(yᵢ - ŷᵢ)²

#     → prediction error


# SECOND PART:

#     α Σβⱼ²

#     → regularization PENALTY


# WHAT IS βⱼ?

#     βⱼ = a coefficient

# Example:

#     β₁ = coefficient of X1
#     β₂ = coefficient of X2
#     β₃ = coefficient of X3

# WHAT IS α?

#     α = regularization strength

# WHAT DOES Σβⱼ² MEAN?

#     Square every coefficient and add them:

#     β₁² + β₂² + β₃² + ... + βₙ²

# Example:

#     β = 10

#     β² = 100

# Large coefficients therefore receive
# a large penalty.

# Ridge says:

#     "Don't just minimize prediction error.
#      Also keep the coefficients small."

# RESULT:

#     Large coefficients
#           ↓
#        SHRINK

#     Usually:

#     β ≠ 0

# So Ridge generally keeps the features but
# reduces their influence.

# --------------------------------------------------
# LASSO — L1 REGULARIZATION
# --------------------------------------------------

# Lasso minimizes:

#     Σ(yᵢ - ŷᵢ)² + α Σ|βⱼ|


# FIRST PART:

#     Σ(yᵢ - ŷᵢ)²

#     → prediction error


# SECOND PART:

#     α Σ|βⱼ|

#     → regularization PENALTY


# WHAT IS |βⱼ|?

#     Absolute value of the coefficient.

# Example:

#     β = -5

#     |β| = 5

# WHAT DOES Σ|βⱼ| MEAN?

#     |β₁| + |β₂| + |β₃| + ... + |βₙ|

# Lasso therefore penalizes the ABSOLUTE SIZE
# of coefficients.

# RESULT:

#     Some coefficients
#           ↓
#         0

# When:

#     β = 0

# the feature contributes nothing to the
# prediction.

# Therefore:

#     LASSO → FEATURE SELECTION

# --------------------------------------------------
# RIDGE vs LASSO
# --------------------------------------------------

# RIDGE — L2:

#     Error + α × Σβ²

#     → Shrinks coefficients
#     → Usually doesn't make them 0
#     → Keeps features
#     → Useful with correlated features


# LASSO — L1:

#     Error + α × Σ|β|

#     → Shrinks coefficients
#     → Can make coefficients exactly 0
#     → Performs feature selection


# --------------------------------------------------
# IMPORTANT SYMBOLS
# --------------------------------------------------

#     y       → actual TARGET

#     ŷ       → predicted TARGET

#     X       → FEATURES / INPUTS

#     β       → COEFFICIENT / SLOPE

#     β₀      → INTERCEPT

#     βⱼ      → coefficient of feature j

#     α       → regularization strength

#     Σ       → summation / add everything

#     |β|     → absolute value of β

#     i       → observation / row number

#     j       → feature / coefficient number


# --------------------------------------------------
# THE CORE INTUITION
# --------------------------------------------------

# WITHOUT REGULARIZATION:

#     Minimize prediction error
#               ↓
#     coefficients can become large


# RIDGE:

#     Minimize prediction error
#               +
#     penalize LARGE coefficients
#               ↓
#     coefficients SHRINK


# LASSO:

#     Minimize prediction error
#               +
#     penalize coefficient size
#               ↓
#     some coefficients can become ZERO
#               ↓
#     FEATURE SELECTION
```
