```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data_dir = Path("/content")  # In Colab, /content is the main working directory, that is why we wrote /content
print((data_dir / "train.csv").exists())
print((data_dir / "test.csv").exists())
print((data_dir / "store.csv").exists())
```

```python
csv_files = list(data_dir.glob("*.csv"))   ####### glob() searches for files matching a pattern, so now it will search for .csv files in the main directory
print(csv_files)  ################## we also used list to make a list of all the csv files only to print  csv_files as an output, thats it  and each item   is a path
```

```python
pd.read_csv("/content/train.csv")
```

```python
required_files = ['store.csv','test.csv','train.csv']
required_paths = [data_dir/ file for file in required_files ]
print(required_paths)
```

```python
for path in required_paths:
  print(path, path.exists())
```

```python
for path in required_paths:
  if not path.exists():
    raise FileNotFoundError(f"missing file:",path)
```

```python
for path in required_paths:
  if not path.is_file():
    raise FileNotFoundError(f"this is not a file:",path)
```
