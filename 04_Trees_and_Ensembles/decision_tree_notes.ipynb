```python
# ============================================================
#              DECISION TREE — THEORY NOTES
# ============================================================


# ============================================================
# 1. WHAT IS A DECISION TREE?
# ============================================================

# Definition:
# A Decision Tree is a supervised ML algorithm that makes
# predictions by repeatedly splitting data using conditions
# on features.
#
# It has a tree-like structure:
#
#              Root
#                |
#             Decision
#            /        \
#          Yes         No
#          |            |
#        Leaf          Leaf
#
# The final leaf gives the prediction.


# ============================================================
# 2. WHY USE DECISION TREES?
# ============================================================

# • Easy to understand and interpret.
# • Can model non-linear relationships.
# • Can automatically learn feature interactions.
# • Does not require feature scaling.
# • Can be used for classification and regression.
# • Can work with numerical and encoded categorical features.
#
# Main disadvantage:
# • Decision Trees can easily overfit if they become too deep.


# ============================================================
# 3. WHERE ARE DECISION TREES USED?
# ============================================================

# Classification:
# • Customer churn
# • Spam detection
# • Loan approval
# • Fraud detection
# • Disease classification
#
# Regression:
# • House price prediction
# • Sales prediction
# • Revenue prediction
# • Demand prediction


# ============================================================
# 4. HOW DOES A DECISION TREE WORK?
# ============================================================

# Step 1:
# Start with the complete dataset.
#
# Step 2:
# Check possible feature splits.
#
# Step 3:
# Measure the quality/impurity of each split.
#
# Step 4:
# Select the best split.
#
# Step 5:
# Repeat the process for child nodes.
#
# Step 6:
# Stop when stopping conditions are reached.
#
# Step 7:
# Final nodes become Leaf Nodes.


# ============================================================
# 5. IMPORTANT TERMINOLOGY
# ============================================================

# Root Node:
# The first/top node of the tree.
#
# Internal Node:
# A node where a decision/split is made.
#
# Leaf Node:
# Final node containing the prediction.
#
# Branch:
# Connection between two nodes.
#
# Split:
# Dividing data according to a feature condition.
#
# Depth:
# Number of levels from the root to a node.


# ============================================================
# 6. CLASSIFICATION vs REGRESSION
# ============================================================

# Decision Tree Classifier:
# Used when the target is categorical.
#
# Example:
# Churn = Yes / No
# Disease = Positive / Negative
#
#
# Decision Tree Regressor:
# Used when the target is continuous numerical data.
#
# Example:
# House Price = ₹50 lakh
# Sales = 12,500


# ============================================================
# 7. DECISION TREE CLASSIFIER
# ============================================================

# Definition:
# A Decision Tree Classifier predicts a categorical target
# by splitting the dataset into increasingly pure groups.
#
# Goal:
# Create child nodes where samples mostly belong to the
# same class.


# ============================================================
# 8. NODE PURITY
# ============================================================

# Pure Node:
# A node where almost/all observations belong to one class.
#
# Example:
#
# Yes  Yes  Yes  Yes
#
# This is completely pure.
#
#
# Impure Node:
#
# Yes  No  Yes  No
#
# This contains mixed classes.


# ============================================================
# 9. IMPURITY
# ============================================================

# Impurity represents how mixed the classes are inside a node.
#
# Low impurity:
#     More pure node
#
# High impurity:
#     More mixed node
#
# Classification commonly uses:
#
#     1. Gini Impurity
#     2. Entropy


# ============================================================
# 10. ENTROPY
# ============================================================

# Definition:
# Entropy measures the uncertainty or impurity in a
# classification node.
#
# Lower entropy:
#     More pure
#
# Higher entropy:
#     More uncertain/mixed


# ------------------------------------------------------------
# ENTROPY FORMULA
# ------------------------------------------------------------

#              C
#             ----
# H(S) = - Σ  p_i log2(p_i)
#             i=1
#
# Where:
#
# p_i = probability/proportion of class i
#
#
# Binary classification:
#
# H(S) = -p1 log2(p1) - p2 log2(p2)


# ------------------------------------------------------------
# ENTROPY SIGNIFICANCE
# ------------------------------------------------------------

# Entropy = 0
#     → Completely pure node.
#
# Entropy = 1
#     → Maximum impurity for binary classification.
#
#
# Example:
#
# 4 Yes, 0 No
#     → Entropy = 0
#
# 2 Yes, 2 No
#     → Entropy = 1


# ============================================================
# 11. INFORMATION GAIN
# ============================================================

# Information Gain measures how much a split reduces entropy.
#
#
# Formula:
#
# Information Gain =
#
# Entropy(parent)
# -
# Weighted Entropy(children)
#
#
# More formally:
#
#                 n_child
# IG = H(parent) - Σ -------- H(child)
#                   n_parent
#
#
# Higher Information Gain:
#     Better split
#
# The tree tries to choose the split with the highest
# information gain when entropy is being used.


# ============================================================
# 12. GINI IMPURITY
# ============================================================

# Definition:
# Gini Impurity measures the probability of incorrectly
# classifying a randomly selected sample if it were labelled
# according to the class distribution of the node.


# ------------------------------------------------------------
# GINI FORMULA
# ------------------------------------------------------------

#              C
#             ----
# Gini = 1 - Σ p_i²
#             i=1
#
# Where:
#
# p_i = proportion of samples belonging to class i


# ------------------------------------------------------------
# GINI SIGNIFICANCE
# ------------------------------------------------------------

# Gini = 0
#     → Completely pure node.
#
# Higher Gini:
#     → More impurity.
#
# For binary classification:
#
# Maximum Gini = 0.5
#
#
# Example:
#
# 100% Yes
#     → Gini = 0
#
# 50% Yes + 50% No
#     → Gini = 0.5


# ============================================================
# 13. GINI vs ENTROPY
# ============================================================

# Gini:
# • Measures impurity.
# • Does not use logarithms.
# • Computationally simpler.
#
# Entropy:
# • Measures uncertainty/information.
# • Uses logarithms.
# • Used with Information Gain.
#
# Both generally produce similar results.
#
# In sklearn:
#
# Gini:
#     criterion = "gini"
#
# Entropy:
#     criterion = "entropy"


# ============================================================
# 14. DECISION TREE REGRESSOR
# ============================================================

# Definition:
# A Decision Tree Regressor predicts a continuous numerical
# value by splitting data into groups containing similar
# target values.
#
# Classification:
#     Make classes more pure.
#
# Regression:
#     Make target values more similar.


# ============================================================
# 15. MSE — MEAN SQUARED ERROR
# ============================================================

# Definition:
# MSE measures the average squared difference between actual
# target values and their mean/predicted value.
#
#
# Formula:
#
#              n
#             ----
# MSE = 1/n * Σ (y_i - y_mean)²
#             i=1
#
# Where:
#
# y_i    = actual value
# y_mean = mean target value in the node
# n      = number of observations


# ------------------------------------------------------------
# WHY MSE IS USED IN REGRESSION TREES
# ------------------------------------------------------------

# The tree tries to create groups where target values are
# close to each other.
#
# Lower MSE:
#     Less variation
#     ↓
#     Better split/node
#
# Higher MSE:
#     More variation
#
#
# Example:
#
# Node A:
# 10, 11, 12, 10
#
# → Low variation → Low MSE
#
#
# Node B:
# 10, 50, 100, 150
#
# → High variation → High MSE


# ============================================================
# 16. PREDICTION IN REGRESSION TREE
# ============================================================

# At a leaf node, the Decision Tree Regressor generally
# predicts the mean target value of the samples in that leaf.
#
#
# Example:
#
# Leaf values:
# 100, 120, 110
#
# Prediction:
#
# (100 + 120 + 110) / 3
#
# = 110


# ============================================================
# 17. OVERFITTING
# ============================================================

# Decision Trees can keep splitting until they memorize
# training data.
#
#
# Very deep tree
#       ↓
# Learns noise and very small patterns
#       ↓
# Very high training performance
#       ↓
# Poor test performance
#       ↓
# Overfitting
#
#
# Example:
#
# Training Accuracy = 100%
# Test Accuracy     = 75%
#
# Possible sign of overfitting.


# ============================================================
# 18. PRUNING
# ============================================================

# Definition:
# Pruning is the process of reducing unnecessary branches
# from a Decision Tree.
#
# Main purpose:
#     Reduce complexity
#     ↓
#     Reduce overfitting
#     ↓
#     Improve generalization


# ============================================================
# 19. TYPES OF PRUNING
# ============================================================

# 1. PRE-PRUNING
# 2. POST-PRUNING


# ------------------------------------------------------------
# PRE-PRUNING
# ------------------------------------------------------------

# Pre-pruning stops the tree from growing too much.
#
# Common methods:
#
# • max_depth
# • min_samples_split
# • min_samples_leaf
# • max_leaf_nodes
# • min_impurity_decrease
#
#
# Example:
#
# max_depth = 5
#
# The tree cannot grow beyond depth 5.


# ------------------------------------------------------------
# POST-PRUNING
# ------------------------------------------------------------

# Post-pruning allows the tree to grow and then removes
# unnecessary branches.
#
# A common sklearn method:
#
# Cost Complexity Pruning
#
# Main parameter:
#
# ccp_alpha


# ============================================================
# 20. COST COMPLEXITY PRUNING
# ============================================================

# Formula:
#
# R_alpha(T) = R(T) + alpha * |T|
#
# Where:
#
# R(T)     = error/impurity of the tree
# alpha    = complexity penalty
# |T|      = number of leaves
#
#
# Higher alpha:
#     Larger penalty for complex trees
#     ↓
#     More pruning
#     ↓
#     Smaller tree


# ============================================================
# 21. IMPORTANT CLASSIFIER PARAMETERS
# ============================================================

# criterion:
# Determines how split quality is measured.
#
# Common:
#     gini
#     entropy
#     log_loss
#
#
# max_depth:
# Maximum depth of the tree.
#
#
# min_samples_split:
# Minimum number of samples required to split a node.
#
#
# min_samples_leaf:
# Minimum number of samples allowed in a leaf.
#
#
# max_features:
# Maximum number of features considered while finding a split.
#
#
# max_leaf_nodes:
# Maximum number of leaf nodes.
#
#
# min_impurity_decrease:
# Minimum impurity reduction required to make a split.
#
#
# ccp_alpha:
# Controls cost-complexity pruning.
#
#
# random_state:
# Controls randomness and makes results reproducible.


# ============================================================
# 22. IMPORTANT REGRESSOR PARAMETERS
# ============================================================

# criterion:
# Determines the split criterion.
#
# Common:
#     squared_error → MSE
#     absolute_error → MAE
#     friedman_mse
#     poisson
#
#
# max_depth:
# Controls maximum tree depth.
#
#
# min_samples_split:
# Minimum samples required to split.
#
#
# min_samples_leaf:
# Minimum samples required in a leaf.
#
#
# max_features:
# Maximum features considered for splitting.
#
#
# max_leaf_nodes:
# Maximum number of leaves.
#
#
# min_impurity_decrease:
# Minimum impurity reduction required.
#
#
# ccp_alpha:
# Controls post-pruning.
#
#
# random_state:
# Ensures reproducibility.


# ============================================================
# 23. IMPORTANT DIFFERENCE — CLASSIFIER vs REGRESSOR
# ============================================================

# Classifier:
#
# Target:
#     Categorical
#
# Main criteria:
#     Gini
#     Entropy
#
# Prediction:
#     Class


# Regressor:
#
# Target:
#     Continuous
#
# Main criterion:
#     MSE / squared error
#
# Prediction:
#     Numerical value


# ============================================================
# 24. FEATURE IMPORTANCE
# ============================================================

# Decision Trees can calculate feature importance based on
# how much each feature contributes to impurity reduction.
#
# Higher importance:
#     Feature contributed more to the tree's splits.
#
# IMPORTANT:
# Feature importance does NOT mean causation.
#
# A feature being important does not prove that it causes
# the target.


# ============================================================
# 25. ADVANTAGES
# ============================================================

# • Easy to understand
# • Easy to visualize
# • Non-linear relationships
# • No feature scaling required
# • Handles feature interactions
# • Works for classification and regression
# • Useful baseline model


# ============================================================
# 26. DISADVANTAGES
# ============================================================

# • Easily overfits
# • Small data changes can create a different tree
# • Deep trees become complex
# • Individual trees may have poor generalization
# • Can be biased toward some features/splits
#
# This is one reason Ensemble methods such as:
#
#     Random Forest
#     Gradient Boosting
#     XGBoost
#
# are commonly used after learning Decision Trees.


# ============================================================
# 27. QUICK FLOW TO REMEMBER
# ============================================================

# DECISION TREE
#      ↓
# Choose feature + threshold
#      ↓
# Evaluate split
#      ↓
# Classification:
#     Gini / Entropy
#      ↓
# Regression:
#     MSE
#      ↓
# Best split
#      ↓
# Repeat
#      ↓
# Leaf
#      ↓
# Prediction


# ============================================================
# 28. ONE-LINE REVISION
# ============================================================

# Decision Tree Classifier:
#     Splits data to make classes purer.
#
# Decision Tree Regressor:
#     Splits data to make target values more similar.
#
# Entropy:
#     Measures uncertainty/impurity.
#
# Information Gain:
#     Measures reduction in entropy.
#
# Gini:
#     Measures class impurity.
#
# MSE:
#     Measures variation of numerical target values.
#
# Pre-pruning:
#     Stop tree growth early.
#
# Post-pruning:
#     Remove unnecessary branches after growth.
#
# max_depth:
#     Controls tree depth.
#
# min_samples_split:
#     Controls when a node can split.
#
# min_samples_leaf:
#     Controls minimum samples in leaves.
#
# ccp_alpha:
#     Controls cost-complexity pruning.
```
