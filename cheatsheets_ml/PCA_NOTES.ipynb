```python
# =========================================================
# PCA — PRINCIPAL COMPONENT ANALYSIS
# =========================================================

# WHAT?
# → Dimensionality reduction technique.
# → Converts many correlated features → fewer "principal components".
# → Keeps as much variance/information as possible.
# → Reduces complexity, noise, memory & computation.

# BASIC FLOW:
# Numerical Data → StandardScaler → PCA → ML Model


# =========================================================
# WHEN TO USE PCA
# =========================================================

# ✅ MANY FEATURES / HIGH-DIMENSIONAL DATA
# → Hundreds/thousands of numerical features.

# ✅ CORRELATED FEATURES
# → Combines correlated variables into fewer components.

# ✅ VISUALIZATION
# → Reduce high-dimensional data → 2D/3D.

# ✅ COMPUTATION
# → Fewer features → potentially faster model training.

# ✅ MODELS AFFECTED BY DIMENSIONALITY
# → SVM, KNN, Logistic Regression, Neural Networks.


# =========================================================
# WHEN NOT TO USE PCA
# =========================================================

# ❌ TREE-BASED MODELS
# → Decision Tree, Random Forest, XGBoost, LightGBM
# → Usually don't need PCA or scaling.

# ❌ VERY FEW FEATURES
# → Little benefit from reducing 5–10 useful features.

# ❌ INTERPRETABILITY IS IMPORTANT
# → PCA components are combinations of original features,
#   so original feature meaning becomes harder to interpret.

# ❌ FEATURES HAVE IMPORTANT INDIVIDUAL MEANING
# → PCA may hide the meaning of the original variables.

# ❌ IF PERFORMANCE GETS WORSE
# → PCA can discard useful information → always compare.


# =========================================================
# PCA + ALGORITHMS
# =========================================================

# Logistic Regression   → ✅ Often useful
# SVM                   → ✅ Often useful
# KNN                   → ✅ Often useful
# Neural Networks       → 🟡 Sometimes
# Linear Regression      → 🟡 Sometimes
# Naive Bayes            → 🟡 Depends on data

# Decision Tree          → ❌ Usually unnecessary
# Random Forest          → ❌ Usually unnecessary
# XGBoost                → ❌ Usually unnecessary
# LightGBM               → ❌ Usually unnecessary


# =========================================================
# BASIC IMPLEMENTATION
# =========================================================

# from sklearn.preprocessing import StandardScaler
# from sklearn.decomposition import PCA

# X_scaled = StandardScaler().fit_transform(X)
# X_pca = PCA(n_components=2).fit_transform(X_scaled)


# # KEEP 95% OF VARIANCE INSTEAD OF FIXING COMPONENTS:
# X_pca = PCA(n_components=0.95).fit_transform(X_scaled)


# =========================================================
# REAL-WORLD WORKFLOW
# =========================================================

# Train/Test Split
#       ↓
# StandardScaler
#       ↓
# PCA
#       ↓
# Model
#       ↓
# Compare with/without PCA
#       ↓
# Keep PCA ONLY if it actually helps


# GOLDEN RULE:
# PCA is NOT automatically better.
# Use it when dimensionality/correlation is a problem,
# then validate whether it improves the model.
```
