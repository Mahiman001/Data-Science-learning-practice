```python
# ╔══════════════════════════════════════════════════════════════════╗
# ║              ML ALGORITHM SELECTION — CHEAT SHEET               ║
# ╚══════════════════════════════════════════════════════════════════╝

# 1️⃣ FIRST IDENTIFY THE PROBLEM
# ──────────────────────────────────────────────────────────────────
# CATEGORY / YES-NO       → CLASSIFICATION
# NUMBER                  → REGRESSION
# NO LABELS / GROUPS      → CLUSTERING
# TOO MANY FEATURES       → DIMENSIONALITY REDUCTION
# TIME-DEPENDENT          → TIME SERIES
# IMAGE                   → COMPUTER VISION
# TEXT                    → NLP


# ══════════════════════════════════════════════════════════════════
# 2️⃣ CLASSIFICATION
# ══════════════════════════════════════════════════════════════════

# LOGISTIC REGRESSION
# → Baseline, simple/linear relationships, interpretable, probabilities
# → Example: churn, disease, loan approval

# DECISION TREE
# → Simple rules + nonlinear relationships + interpretable
# → Small/medium tabular data

# RANDOM FOREST
# → Strong general-purpose tabular classifier
# → Nonlinear data, robust, less overfitting than single tree
# → Example: fraud, churn

# XGBOOST ⭐
# → Complex nonlinear tabular data + strong predictive performance
# → Usually one of the first serious models to test
# → Example: fraud, churn, phishing

# LIGHTGBM ⭐
# → Large tabular datasets + speed
# → Often faster than XGBoost on large data

# SVM
# → Small/medium datasets + complex decision boundaries
# → Good for high-dimensional data

# KNN
# → Similarity-based prediction + small datasets
# → Avoid for very large datasets

# NAIVE BAYES
# → Text + high-dimensional sparse data
# → Very fast
# → Example: spam/email classification

# TEXT
# → Naive Bayes / Logistic Regression
# → Advanced NLP → Transformers

# IMAGE
# → CNN / Vision Transformer


# ══════════════════════════════════════════════════════════════════
# 3️⃣ REGRESSION
# ══════════════════════════════════════════════════════════════════

# LINEAR REGRESSION
# → Linear relationship + excellent baseline
# → Example: simple house-price prediction

# RIDGE
# → Linear + correlated features
# → Controls large coefficients

# LASSO
# → Feature selection
# → Pushes some coefficients → 0

# ELASTIC NET
# → Correlated features + feature selection
# → Ridge + Lasso combination

# DECISION TREE
# → Nonlinear relationships + rule-based prediction

# RANDOM FOREST
# → Strong general-purpose tabular regression
# → Robust + nonlinear

# XGBOOST ⭐
# → Complex nonlinear tabular regression
# → Often very strong performance

# LIGHTGBM ⭐
# → Large datasets + speed


# REAL EXAMPLES:
# House price → Linear/Ridge → RF → XGBoost/LightGBM
# Sales       → XGBoost/LightGBM
# Salary      → Linear/Ridge → RF/XGBoost
# Demand      → XGBoost/LightGBM or time-series models


# ══════════════════════════════════════════════════════════════════
# 4️⃣ CLUSTERING
# ══════════════════════════════════════════════════════════════════

# K-MEANS
# → Customer segmentation / grouping
# → Fast + simple
# → Best when clusters are roughly spherical
# → Must choose K

# DBSCAN
# → Irregular-shaped clusters + noise/outliers
# → No need to specify number of clusters
# → Example: geographic/anomaly detection
# → Weak when cluster densities vary greatly

# HIERARCHICAL
# → Small/medium datasets + want cluster hierarchy
# → Dendrogram
# → Avoid huge datasets


# ══════════════════════════════════════════════════════════════════
# 5️⃣ PCA
# ══════════════════════════════════════════════════════════════════

# PCA
# → Reduce many features → fewer features
# → Visualization / correlated features / speed
# → NOT normally a prediction algorithm

# Typical:
# Data → StandardScaler → PCA → ML Model


# ══════════════════════════════════════════════════════════════════
# 6️⃣ TIME SERIES
# ══════════════════════════════════════════════════════════════════

# Examples:
# Sales / stocks / traffic / electricity / weather

# Options:
# → Lag features + Linear/Tree models
# → XGBoost / LightGBM
# → ARIMA / SARIMA / Prophet
# → LSTM / Transformers

# ⚠️ Time order matters.
# Don't randomly shuffle time-series data like normal ML.


# ══════════════════════════════════════════════════════════════════
# 7️⃣ DEEP LEARNING — WHEN?
# ══════════════════════════════════════════════════════════════════

# TABULAR       → Usually XGBoost / LightGBM first
# IMAGE         → CNN / Vision Transformer
# TEXT          → Transformers
# AUDIO         → CNN / Transformers
# SEQUENCES     → LSTM / GRU / Transformers

# Small tabular data → Tree models often make more sense
# Large/unstructured → Deep Learning becomes more attractive


# ══════════════════════════════════════════════════════════════════
# 8️⃣ QUICK DECISION
# ══════════════════════════════════════════════════════════════════

#               WHAT ARE YOU PREDICTING?
#                        │
#        ┌───────────────┼────────────────┐
#        ↓               ↓                ↓
#    CATEGORY          NUMBER          NO LABEL
#        ↓               ↓                ↓
#  CLASSIFICATION    REGRESSION      CLUSTERING
#        ↓               ↓                ↓
#  Logistic          Linear          K-Means
#  RF                Ridge           DBSCAN
#  XGBoost ⭐        Lasso           Hierarchical
#  LightGBM
#  SVM
#  Naive Bayes


# ══════════════════════════════════════════════════════════════════
# 9️⃣ WHAT TO TRY FIRST
# ══════════════════════════════════════════════════════════════════

# TABULAR CLASSIFICATION:
# Logistic → RF → XGBoost → LightGBM → SVM

# TABULAR REGRESSION:
# Linear/Ridge → RF → XGBoost → LightGBM

# TEXT:
# Naive Bayes → Logistic → Transformer

# CLUSTERING:
# K-Means → DBSCAN → Hierarchical

# IMAGE:
# CNN / Vision Transformer

# SEQUENCE:
# LSTM / Transformer


# ══════════════════════════════════════════════════════════════════
# 🔟 REAL-WORLD SELECTION RULE
# ══════════════════════════════════════════════════════════════════

# DON'T ASK:
# ❌ "What's the best algorithm?"

# ASK:
# ✅ "Which algorithms fit my data and constraints?"

# Consider:
# → Dataset size
# → Feature types
# → Linear/nonlinear relationships
# → Missing values/outliers
# → Class imbalance
# → Accuracy/performance
# → Interpretability
# → Training speed
# → Prediction speed
# → Memory/compute
# → Deployment requirements


# ══════════════════════════════════════════════════════════════════
# 1️⃣1️⃣ PROJECT WORKFLOW
# ══════════════════════════════════════════════════════════════════

# Problem
#  ↓
# Target + problem type
#  ↓
# Understand data
#  ↓
# EDA + preprocessing
#  ↓
# Baseline
#  ↓
# Suitable algorithms
#  ↓
# Cross-validation
#  ↓
# Hyperparameter tuning
#  ↓
# Compare metrics
#  ↓
# Check overfitting
#  ↓
# Final model
#  ↓
# Save → Deploy → Monitor


# CLASSIFICATION:
# Accuracy / Precision / Recall / F1 / ROC-AUC / PR-AUC

# REGRESSION:
# MAE / RMSE / R²


# ╔══════════════════════════════════════════════════════════════════╗
# ║ GOLDEN RULE:                                                    ║
# ║ Problem → Data → Baseline → Candidates → Validation → Tuning   ║
# ║ → Comparison → Final Model                                     ║
# ╚══════════════════════════════════════════════════════════════════╝
```
