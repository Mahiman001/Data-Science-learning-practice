```python
# # =========================================================
# # CLUSTERING TECHNIQUES — QUICK REVISION SHEET
# # =========================================================

# # 1. K-MEANS
# # ---------------------------------------------------------
# # WHAT:
# # Divides data into K clusters based on distance from centroids.
# #
# # WHEN:
# # - You know/estimate the number of clusters
# # - Clusters are roughly spherical and similarly sized
# #
# # WHERE:
# # - Customer segmentation
# # - Market segmentation
# # - Grouping similar data
# #
# # WHY:
# # - Simple
# # - Fast
# # - Good for large datasets
# #
# # KEY PARAMETER:
# # n_clusters
# #
# # CODE:
# from sklearn.cluster import KMeans
# model = KMeans(n_clusters=3, random_state=42)
# labels = model.fit_predict(X)


# # 2. HIERARCHICAL / AGGLOMERATIVE
# # ---------------------------------------------------------
# # WHAT:
# # Builds a hierarchy/tree of clusters by repeatedly
# # merging the closest clusters.
# #
# # WHEN:
# # - You want to understand relationships between clusters
# # - You don't know the ideal number of clusters initially
# #
# # WHERE:
# # - Biology / gene clustering
# # - Customer segmentation
# # - Document clustering
# #
# # WHY:
# # - Produces a dendrogram
# # - Shows cluster hierarchy
# #
# # KEY PARAMETERS:
# # n_clusters, linkage
# #
# # CODE:
# from sklearn.cluster import AgglomerativeClustering
# model = AgglomerativeClustering(n_clusters=3, linkage="ward")
# labels = model.fit_predict(X)


# # 3. DBSCAN
# # ---------------------------------------------------------
# # WHAT:
# # Density-based clustering that groups dense regions
# # and identifies isolated points as noise.
# #
# # WHEN:
# # - Number of clusters is unknown
# # - Data has irregular/non-spherical shapes
# # - Dataset contains outliers/noise
# #
# # WHERE:
# # - Spatial/geographical data
# # - Anomaly/noise detection
# # - Irregular-shaped clusters
# #
# # WHY:
# # - No need to specify number of clusters
# # - Detects outliers automatically
# #
# # KEY PARAMETERS:
# # eps, min_samples
# #
# # CODE:
# from sklearn.cluster import DBSCAN
# model = DBSCAN(eps=0.2, min_samples=5)
# labels = model.fit_predict(X)


# # 4. GAUSSIAN MIXTURE MODEL (GMM)
# # ---------------------------------------------------------
# # WHAT:
# # Probabilistic clustering using multiple Gaussian
# # distributions.
# #
# # WHEN:
# # - Clusters overlap
# # - You want probability of cluster membership
# # - Clusters are approximately Gaussian/elliptical
# #
# # WHERE:
# # - Customer segmentation
# # - Pattern recognition
# # - Density estimation
# #
# # WHY:
# # - Soft clustering
# # - Gives probability instead of only hard labels
# #
# # KEY PARAMETER:
# # n_components
# #
# # CODE:
# from sklearn.mixture import GaussianMixture
# model = GaussianMixture(n_components=3, random_state=42)
# labels = model.fit_predict(X)


# # 5. MEAN SHIFT
# # ---------------------------------------------------------
# # WHAT:
# # Finds dense regions by moving points toward areas
# # of higher data density.
# #
# # WHEN:
# # - Number of clusters is unknown
# # - You want density-based clustering
# #
# # WHERE:
# # - Image processing
# # - Computer vision
# # - Pattern detection
# #
# # WHY:
# # - No need to specify number of clusters
# #
# # KEY PARAMETER:
# # bandwidth
# #
# # CODE:
# from sklearn.cluster import MeanShift
# model = MeanShift()
# labels = model.fit_predict(X)


# # 6. SPECTRAL CLUSTERING
# # ---------------------------------------------------------
# # WHAT:
# # Treats data as a graph and uses eigenvalues/eigenvectors
# # to identify clusters.
# #
# # WHEN:
# # - Clusters have complex/non-convex shapes
# #
# # WHERE:
# # - Image segmentation
# # - Graph/network clustering
# # - Complex datasets
# #
# # WHY:
# # - Can find clusters that K-Means cannot
# #
# # KEY PARAMETER:
# # n_clusters
# #
# # CODE:
# from sklearn.cluster import SpectralClustering
# model = SpectralClustering(n_clusters=3, random_state=42)
# labels = model.fit_predict(X)


# # 7. OPTICS
# # ---------------------------------------------------------
# # WHAT:
# # Density-based algorithm related to DBSCAN but better
# # at handling datasets with different densities.
# #
# # WHEN:
# # - Cluster density varies
# # - DBSCAN's single eps value doesn't work well
# #
# # WHERE:
# # - Spatial data
# # - Large/noisy datasets
# #
# # WHY:
# # - Handles varying density better than DBSCAN
# #
# # KEY PARAMETERS:
# # min_samples, max_eps
# #
# # CODE:
# from sklearn.cluster import OPTICS
# model = OPTICS(min_samples=5)
# labels = model.fit_predict(X)


# # 8. BIRCH
# # ---------------------------------------------------------
# # WHAT:
# # Incrementally builds a clustering tree to efficiently
# # handle large datasets.
# #
# # WHEN:
# # - Dataset is very large
# # - Memory/computation efficiency matters
# #
# # WHERE:
# # - Large-scale clustering
# # - Streaming/large datasets
# #
# # WHY:
# # - Efficient for very large datasets
# #
# # KEY PARAMETERS:
# # threshold, n_clusters
# #
# # CODE:
# from sklearn.cluster import Birch
# model = Birch(n_clusters=3)
# labels = model.fit_predict(X)


# # =========================================================
# # ONE-LINE COMPARISON
# # =========================================================

# K-Means      → Centroid-based
# Hierarchical → Tree-based
# DBSCAN       → Density-based + noise detection
# GMM          → Probability-based / soft clustering
# Mean Shift   → Density peak-based
# Spectral     → Graph-based
# OPTICS       → Density-based + varying density
# BIRCH        → Large-scale efficient clustering


# # =========================================================
# # WHICH ONE WHEN?
# # =========================================================

# Know number of clusters
#         ↓
#     K-MEANS

# Unknown number + irregular shapes + outliers
#         ↓
#       DBSCAN

# Different cluster densities
#         ↓
#       OPTICS

# Need hierarchy / dendrogram
#         ↓
#    HIERARCHICAL

# Overlapping clusters + probabilities
#         ↓
#        GMM

# Complex/non-convex graph-like structure
#         ↓
#     SPECTRAL

# Very large dataset
#         ↓
#       BIRCH

# Density-based + don't know cluster count
#         ↓
#    MEAN SHIFT
```
