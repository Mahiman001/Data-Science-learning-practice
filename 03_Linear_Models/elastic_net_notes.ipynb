```python
# # ============================================================
# # ELASTIC NET REGRESSION — COMPLETE NOTES
# # ============================================================


# # 1. WHAT IS ELASTIC NET?
# # ------------------------------------------------------------
# # Elastic Net is a regularized linear regression algorithm.
# #
# # It combines:
# #   1. Ridge Regression  -> L2 regularization
# #   2. Lasso Regression  -> L1 regularization
# #
# # So:
# #
# #       Elastic Net = Ridge + Lasso
# #
# # It is useful when we have:
# #   - Many features
# #   - Correlated features
# #   - Risk of overfitting
# #   - Need for some feature selection


# # ============================================================
# # 2. WHY DO WE NEED ELASTIC NET?
# # ============================================================

# # Normal Linear Regression tries to minimize:
# #
# #       Sum of Squared Errors
# #
# # Problem:
# # If we have many features or highly correlated features,
# # the model can become too complex and overfit.
# #
# # Regularization adds a penalty for large coefficients.
# #
# # Elastic Net uses BOTH:
# #
# #       L1 penalty -> can make coefficients exactly 0
# #       L2 penalty -> shrinks coefficients toward 0
# #
# # This gives us the advantages of both Ridge and Lasso.


# # ============================================================
# # 3. QUICK COMPARISON
# # ============================================================

# # Linear Regression
# # -----------------
# # No regularization
# # Coefficients can become large
# # More chance of overfitting


# # Ridge Regression
# # ----------------
# # Uses L2 regularization
# # Shrinks coefficients
# # Usually keeps all features
# #
# # Penalty:
# #       alpha * sum(coefficient^2)


# # Lasso Regression
# # ----------------
# # Uses L1 regularization
# # Shrinks coefficients
# # Can make coefficients exactly 0
# # Therefore performs feature selection
# #
# # Penalty:
# #       alpha * sum(abs(coefficient))


# # Elastic Net
# # -----------
# # Uses BOTH L1 and L2
# # Can perform feature selection
# # Also handles correlated features better than pure Lasso


# # ============================================================
# # 4. MAIN IDEA
# # ============================================================

# # Suppose our features are:
# #
# # Age
# # Income
# # Experience
# # House_Size
# # Distance
# #
# # Each feature gets a coefficient:
# #
# #       y = b0 + b1(Age) + b2(Income) + ...
# #
# # Elastic Net controls these coefficients.
# #
# # Weak/unimportant features may have their coefficients
# # reduced heavily or even become 0.
# #
# # Important features retain larger coefficients.


# # ============================================================
# # 5. ELASTIC NET FORMULA
# # ============================================================

# # Objective function:
# #
# #   Minimize:
# #
# #   SSE + alpha * [
# #                   l1_ratio * sum(|coefficients|)
# #                   +
# #                   (1-l1_ratio) * sum(coefficients^2)
# #                 ]
# #
# #
# # Where:
# #
# # alpha      = overall regularization strength
# #
# # l1_ratio   = controls the mixture of Lasso and Ridge
# #
# #
# # Important:
# #
# # l1_ratio = 1
# #       -> Pure Lasso
# #
# # l1_ratio = 0
# #       -> Pure Ridge
# #
# # 0 < l1_ratio < 1
# #       -> Elastic Net combination


# # ============================================================
# # 6. ALPHA
# # ============================================================

# # alpha controls HOW STRONG the regularization is.
# #
# # Small alpha:
# #       Less regularization
# #       Model behaves more like Linear Regression
# #
# # Large alpha:
# #       Stronger regularization
# #       Coefficients shrink more
# #       Can reduce overfitting
# #
# #
# # Example:
# #
# # alpha = 0.01   -> weak regularization
# # alpha = 1      -> moderate
# # alpha = 10     -> strong
# #
# # These are examples, NOT universal best values.


# # ============================================================
# # 7. L1_RATIO
# # ============================================================

# # l1_ratio controls the balance between Lasso and Ridge.
# #
# # Example:
# #
# # l1_ratio = 0.2
# #
# #       20% Lasso
# #       80% Ridge
# #
# #
# # l1_ratio = 0.5
# #
# #       50% Lasso
# #       50% Ridge
# #
# #
# # l1_ratio = 0.8
# #
# #       80% Lasso
# #       20% Ridge
# #
# #
# # l1_ratio = 1
# #
# #       100% Lasso
# #
# #
# # l1_ratio = 0
# #
# #       100% Ridge


# # ============================================================
# # 8. WHY NOT JUST USE LASSO?
# # ============================================================

# # Lasso can struggle when features are highly correlated.
# #
# # Example:
# #
# #       House_Size
# #       Number_of_Rooms
# #
# # Both may contain similar information.
# #
# # Lasso may select one feature and reduce the other toward zero.
# #
# # Elastic Net combines L1 + L2, which can provide more stable
# # behavior with correlated features.


# # ============================================================
# # 9. WHY NOT JUST USE RIDGE?
# # ============================================================

# # Ridge shrinks coefficients but generally does not perform
# # feature selection.
# #
# # If we have 100 features and only 20 are useful,
# # Ridge can still retain coefficients for many/all features.
# #
# # Elastic Net can push some coefficients exactly to 0
# # because of its L1 component.


# # ============================================================
# # 10. WHEN SHOULD WE USE ELASTIC NET?
# # ============================================================

# # Use Elastic Net when:
# #
# # 1. You have many features
# #
# # 2. Features are highly correlated
# #
# # 3. You suspect overfitting
# #
# # 4. You want regularization
# #
# # 5. You also want some feature selection
# #
# # 6. Linear relationships are reasonably appropriate
# #
# #
# # Especially useful for datasets with:
# #       High dimensionality
# #       Multicollinearity
# #       Many potentially irrelevant features


# # ============================================================
# # 11. WHEN SHOULD WE NOT USE IT?
# # ============================================================

# # Elastic Net is still a LINEAR model.
# #
# # If the relationship is strongly nonlinear,
# # tree-based models such as:
# #
# #       Random Forest
# #       XGBoost
# #       LightGBM
# #
# # may be more appropriate.
# #
# # Also, regularization is not automatically better.
# # It should be validated using appropriate evaluation.


# # ============================================================
# # 12. IMPORTANT: FEATURE SCALING
# # ============================================================

# # Elastic Net is sensitive to feature scale because
# # regularization acts directly on coefficients.
# #
# # Example:
# #
# # Income:
# #       20,000 -> 100,000
# #
# # Age:
# #       18 -> 60
# #
# # Their scales are very different.
# #
# # Therefore we usually STANDARDIZE numerical features.
# #
# # StandardScaler:
# #
# #       z = (x - mean) / standard_deviation
# #
# #
# # IMPORTANT:
# # Fit the scaler ONLY on training data.
# #
# # Using Pipeline helps prevent data leakage.


# # ============================================================
# # 13. BASIC SKLEARN SYNTAX
# # ============================================================

# from sklearn.linear_model import ElasticNet

# model = ElasticNet(
#     alpha=1.0,
#     l1_ratio=0.5,
#     max_iter=10000
# )

# model.fit(X_train, y_train)

# y_pred = model.predict(X_test)


# # ============================================================
# # 14. PARAMETERS
# # ============================================================

# # alpha
# #       Controls overall regularization strength.
# #
# # l1_ratio
# #       Controls L1 vs L2 contribution.
# #
# # max_iter
# #       Maximum number of optimization iterations.
# #
# # random_state
# #       Can be used for reproducibility in relevant settings.


# # ============================================================
# # 15. ELASTIC NET WITH STANDARDIZATION
# # ============================================================

# from sklearn.pipeline import Pipeline
# from sklearn.preprocessing import StandardScaler
# from sklearn.linear_model import ElasticNet

# model = Pipeline([
#     ("scaler", StandardScaler()),
#     ("elasticnet", ElasticNet(
#         alpha=1.0,
#         l1_ratio=0.5,
#         max_iter=10000
#     ))
# ])

# model.fit(X_train, y_train)

# y_pred = model.predict(X_test)


# # ============================================================
# # 16. EVALUATION
# # ============================================================

# from sklearn.metrics import mean_absolute_error
# from sklearn.metrics import mean_squared_error, r2_score

# mae = mean_absolute_error(y_test, y_pred)

# rmse = mean_squared_error(
#     y_test,
#     y_pred
# ) ** 0.5

# r2 = r2_score(y_test, y_pred)

# print("MAE:", mae)
# print("RMSE:", rmse)
# print("R2:", r2)


# # MAE
# # ----
# # Average absolute prediction error.
# #
# # Lower = better


# # RMSE
# # ----
# # Penalizes large errors more strongly than MAE.
# #
# # Lower = better


# # R²
# # --
# # Measures how much variance in the target is explained
# # by the model.
# #
# # Higher is generally better.


# # ============================================================
# # 17. GETTING COEFFICIENTS
# # ============================================================

# # If using ElasticNet directly:
# #
# model = ElasticNet(
#     alpha=1.0,
#     l1_ratio=0.5,
#     max_iter=10000
# )

# model.fit(X_train, y_train)

# print(model.coef_)
# print(model.intercept_)


# # coef_
# #       -> slopes / weights of the features
# #
# # intercept_
# #       -> b0 in:
# #
# #       y = b0 + b1x1 + b2x2 + ...


# # ============================================================
# # 18. FEATURE SELECTION
# # ============================================================

# # L1 regularization can make coefficients exactly zero.
# #
# # Example:
# #
# # Feature A -> coefficient = 5.21
# # Feature B -> coefficient = 0.03
# # Feature C -> coefficient = 0
# # Feature D -> coefficient = -2.17
# #
# # Feature C has effectively been removed from the model.
# #
# # Therefore:
# #
# #       coefficient == 0
# #
# # can indicate that the feature was not selected by the model.


# # ============================================================
# # 19. ELASTIC NET + CROSS VALIDATION
# # ============================================================

# # We normally don't manually guess alpha and l1_ratio.
# #
# # We can test multiple combinations using cross-validation.


# from sklearn.linear_model import ElasticNetCV

# model = ElasticNetCV(
#     alphas=[0.001, 0.01, 0.1, 1, 10],
#     l1_ratio=[0.1, 0.3, 0.5, 0.7, 0.9],
#     cv=5,
#     max_iter=10000
# )

# model.fit(X_train, y_train)

# y_pred = model.predict(X_test)

# print("Best alpha:", model.alpha_)
# print("Best l1_ratio:", model.l1_ratio_)


# # ============================================================
# # 20. WHY USE ELASTICNETCV?
# # ============================================================

# # Instead of manually deciding:
# #
# #       alpha = ?
# #       l1_ratio = ?
# #
# # ElasticNetCV tests different combinations using
# # cross-validation and selects suitable values based on
# # validation performance.
# #
# # This is generally preferable when tuning the model.


# # ============================================================
# # 21. ELASTIC NET WITH PIPELINE + CV
# # ============================================================

# from sklearn.pipeline import Pipeline
# from sklearn.preprocessing import StandardScaler
# from sklearn.linear_model import ElasticNetCV

# model = Pipeline([
#     ("scaler", StandardScaler()),
#     ("elasticnet", ElasticNetCV(
#         alphas=[0.001, 0.01, 0.1, 1, 10],
#         l1_ratio=[0.1, 0.5, 0.9],
#         cv=5,
#         max_iter=10000
#     ))
# ])

# model.fit(X_train, y_train)

# y_pred = model.predict(X_test)


# # ============================================================
# # 22. IMPORTANT PIPELINE IDEA
# # ============================================================

# # Pipeline:
# #
# #       Raw Data
# #           ↓
# #       StandardScaler
# #           ↓
# #       Elastic Net
# #           ↓
# #       Prediction
# #
# #
# # Why?
# #
# # 1. Scaling happens correctly.
# # 2. Prevents data leakage.
# # 3. Keeps preprocessing + model together.
# # 4. Easier to deploy.


# # ============================================================
# # 23. RIDGE vs LASSO vs ELASTIC NET
# # ============================================================

# # Linear Regression
# #       No regularization
# #       No feature selection
# #
# #
# # Ridge
# #       L2
# #       Shrinks coefficients
# #       Usually keeps features
# #
# #
# # Lasso
# #       L1
# #       Shrinks coefficients
# #       Can set coefficients to 0
# #       Feature selection
# #
# #
# # Elastic Net
# #       L1 + L2
# #       Shrinking + feature selection
# #       Useful with correlated features


# # ============================================================
# # 24. SIMPLE INTUITION
# # ============================================================

# # Think of regularization as controlling model complexity.
# #
# # Linear Regression:
# #       "Use the data freely."
# #
# # Ridge:
# #       "Keep all features, but don't let coefficients
# #        become unnecessarily large."
# #
# # Lasso:
# #       "Remove some features if possible."
# #
# # Elastic Net:
# #       "Shrink coefficients AND remove some features,
# #        while handling correlated features better."


# # ============================================================
# # 25. KEY POINTS TO REMEMBER
# # ============================================================

# # 1. Elastic Net = L1 + L2 regularization.
# #
# # 2. L1 comes from Lasso.
# #
# # 3. L2 comes from Ridge.
# #
# # 4. alpha controls overall regularization strength.
# #
# # 5. l1_ratio controls Lasso/Ridge balance.
# #
# # 6. l1_ratio = 1 -> Lasso.
# #
# # 7. l1_ratio = 0 -> Ridge.
# #
# # 8. L1 can make coefficients exactly 0.
# #
# # 9. Therefore Elastic Net can perform feature selection.
# #
# # 10. L2 helps with correlated features.
# #
# # 11. Feature scaling is usually important.
# #
# # 12. Use Pipeline to combine scaling + model.
# #
# # 13. Use ElasticNetCV to tune alpha and l1_ratio.
# #
# # 14. Evaluate using MAE, RMSE and R².
# #
# # 15. Elastic Net is still a linear model.


# # ============================================================
# # 26. OUR ML PROGRESSION
# # ============================================================

# # Linear Regression
# #        ↓
# # Ridge Regression
# #        ↓
# # Lasso Regression
# #        ↓
# # Elastic Net
# #        ↓
# # Tree-based Models
# #        ↓
# # XGBoost / LightGBM
# #
# # Elastic Net completes our regularized linear regression
# # section.
```
