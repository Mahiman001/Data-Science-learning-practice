```python


━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
BEFORE ANY TEST → CHECK ASSUMPTIONS FIRST
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

1. SHAPIRO-WILK TEST (Normality Check)
   ├── WHEN : Before any parametric test (t-test, ANOVA)
   ├── WHY  : Check if data follows a bell curve
   ├── HOW  : p > 0.05 → Normal ✅ | p < 0.05 → Not Normal ❌
   └── CODE :
            stat, p = stats.shapiro(data)
            # Example: Checking if exam scores are normally distributed

2. LEVENE'S TEST (Equal Variance Check)
   ├── WHEN : Before t-test or ANOVA (after normality confirmed)
   ├── WHY  : Check if groups have similar spread/variance
   ├── HOW  : p > 0.05 → Equal Variance ✅ | p < 0.05 → Unequal ❌
   └── CODE :
            stat, p = stats.levene(group1, group2, group3)
            # Example: Do all 3 classes have similar score spreads?

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
COMPARING MEANS
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

3. ONE SAMPLE T-TEST
   ├── WHEN : Compare ONE group mean against a KNOWN value
   ├── WHY  : Is our sample different from a standard/target?
   ├── HOW  : p < 0.05 → Mean IS different from target
   ├── NEED : Normal data, one group
   └── CODE :
            stat, p = stats.ttest_1samp(data, popmean=70)
            # Example: Is our class average different from national avg of 70?

4. INDEPENDENT T-TEST (Two Sample)
   ├── WHEN : Compare means of TWO DIFFERENT/UNRELATED groups
   ├── WHY  : Is Group A different from Group B?
   ├── HOW  : p < 0.05 → Groups ARE significantly different
   ├── NEED : Normal data, equal variance, independent groups
   └── CODE :
            stat, p = stats.ttest_ind(group_a, group_b)
            # Example: Do boys and girls score differently?

5. PAIRED T-TEST
   ├── WHEN : Compare SAME group measured TWICE (before/after)
   ├── WHY  : Did something change after treatment/intervention?
   ├── HOW  : p < 0.05 → Significant change occurred
   ├── NEED : Normal data, same subjects measured twice
   └── CODE :
            stat, p = stats.ttest_rel(before, after)
            # Example: Did training improve employee performance?

6. WELCH'S T-TEST
   ├── WHEN : Like independent t-test BUT variances are UNEQUAL
   ├── WHY  : Safer version when spread between groups differs
   ├── HOW  : p < 0.05 → Groups ARE significantly different
   ├── NEED : Normal data, unequal variance, independent groups
   └── CODE :
            stat, p = stats.ttest_ind(group_a, group_b, equal_var=False)
            # Example: Comparing salaries (very different spreads)

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
COMPARING 3+ GROUPS
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

7. ONE-WAY ANOVA
   ├── WHEN : Compare means of 3 or MORE groups (one factor)
   ├── WHY  : Are ANY of the groups different from each other?
   ├── HOW  : p < 0.05 → At least one group is different
   ├── NEED : Normal data, equal variance, independent groups
   ├── NOTE : Only tells you IF difference exists, not WHERE
   └── CODE :
            stat, p = stats.f_oneway(group_a, group_b, group_c)
            # Example: Which teaching method (A/B/C) works best?

8. TWO-WAY ANOVA
   ├── WHEN : Testing effect of TWO factors simultaneously
   ├── WHY  : Check main effects AND interaction between factors
   ├── HOW  : p < 0.05 for each factor → That factor matters
   ├── NEED : Normal data, equal variance, balanced design
   └── CODE :
            model = ols('Score ~ C(Method) + C(Gender) + C(Method):C(Gender)',
                        data=df).fit()
            table = sm.stats.anova_lm(model, typ=2)
            # Example: Effect of teaching method AND gender on scores
            # Also checks: Does gender affect which method works better?

9. REPEATED MEASURES ANOVA
   ├── WHEN : Same subjects measured 3+ times (multiple timepoints)
   ├── WHY  : Track change over time within same group
   ├── HOW  : p < 0.05 → Significant change across time points
   ├── NEED : Normal data, sphericity assumption
   └── CODE :
            # Using pingouin library
            import pingouin as pg
            result = pg.rm_anova(data=df, dv='Score',
                                 within='Time', subject='Subject')
            # Example: Patient recovery scores at week 1, 2, 3, 4

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
POST-HOC TESTS (After ANOVA - Finding WHERE the difference is)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

10. TUKEY'S HSD
    ├── WHEN : After ANOVA is significant, equal group sizes
    ├── WHY  : Find WHICH specific pairs are different
    ├── HOW  : reject=True → That pair IS significantly different
    └── CODE :
             tukey = pairwise_tukeyhsd(endog=df['Score'],
                                       groups=df['Method'], alpha=0.05)
             # Example: ANOVA says A/B/C differ → Tukey says C≠A, C≠B, A=B

11. BONFERRONI CORRECTION
    ├── WHEN : Multiple comparisons, want strict control
    ├── WHY  : Prevents false positives from multiple testing
    ├── HOW  : New alpha = 0.05 / number of tests
    └── CODE :
             from statsmodels.stats.multitest import multipletests
             reject, p_corrected, _, _ = multipletests(p_values, method='bonferroni')
             # Example: Testing 10 drugs → use 0.005 as threshold instead of 0.05

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
NON-PARAMETRIC TESTS (When data is NOT normal)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

12. MANN-WHITNEY U TEST  ← Alternative to Independent T-Test
    ├── WHEN : Compare 2 independent groups, data NOT normal
    ├── WHY  : Works on ranks instead of actual values
    ├── HOW  : p < 0.05 → Groups are significantly different
    └── CODE :
             stat, p = stats.mannwhitneyu(group_a, group_b)
             # Example: Comparing satisfaction ratings (1-5 scale)

13. WILCOXON SIGNED-RANK  ← Alternative to Paired T-Test
    ├── WHEN : Before/after comparison, data NOT normal
    ├── WHY  : Non-parametric version of paired t-test
    ├── HOW  : p < 0.05 → Significant change occurred
    └── CODE :
             stat, p = stats.wilcoxon(before, after)
             # Example: Pain levels before/after medication (skewed data)

14. KRUSKAL-WALLIS TEST  ← Alternative to One-Way ANOVA
    ├── WHEN : Compare 3+ groups, data NOT normal
    ├── WHY  : Non-parametric version of ANOVA (uses ranks)
    ├── HOW  : p < 0.05 → At least one group differs
    └── CODE :
             stat, p = stats.kruskal(group_a, group_b, group_c)
             # Example: Customer ratings across 3 store locations

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
RELATIONSHIP BETWEEN VARIABLES
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

15. PEARSON CORRELATION
    ├── WHEN : Check relationship between 2 numeric variables
    ├── WHY  : Are they linearly related? How strongly?
    ├── HOW  : r close to +1/-1 = Strong | r close to 0 = Weak
    ├── NEED : Both variables must be normal and continuous
    └── CODE :
             r, p = stats.pearsonr(x, y)
             # Example: Does study time correlate with test score?
             # r = 0.85, p < 0.05 → Strong positive relationship

16. SPEARMAN CORRELATION  ← Non-parametric version
    ├── WHEN : Relationship between variables, NOT normal or ordinal data
    ├── WHY  : Works on ranks, more robust to outliers
    ├── HOW  : Same interpretation as Pearson
    └── CODE :
             r, p = stats.spearmanr(x, y)
             # Example: Does movie ranking correlate with revenue?

17. CHI-SQUARE TEST
    ├── WHEN : Both variables are CATEGORICAL (not numbers)
    ├── WHY  : Is there a relationship between two categories?
    ├── HOW  : p < 0.05 → Variables ARE related/associated
    └── CODE :
             contingency_table = pd.crosstab(df['Gender'], df['Preference'])
             stat, p, dof, expected = stats.chi2_contingency(contingency_table)
             # Example: Is smoking related to gender? (Yes/No vs Male/Female)

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
PREDICTION TESTS
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

18. LINEAR REGRESSION
    ├── WHEN : Predict a NUMBER using one or more variables
    ├── WHY  : Build equation to predict outcomes
    ├── HOW  : p < 0.05 for variable → It's a useful predictor
    └── CODE :
             model = ols('Score ~ StudyHours + Sleep', data=df).fit()
             print(model.summary())
             # Example: Predict exam score from study hours and sleep

19. LOGISTIC REGRESSION
    ├── WHEN : Predict a CATEGORY (Yes/No, Pass/Fail)
    ├── WHY  : Classify outcomes based on predictors
    ├── HOW  : Outputs probability (0 to 1)
    └── CODE :
             from sklearn.linear_model import LogisticRegression
             model = LogisticRegression()
             model.fit(X_train, y_train)
             # Example: Predict if student passes/fails based on study hours

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
QUICK DECISION GUIDE
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

How many groups?
│
├── ONE GROUP
│   └── vs known value? → One Sample T-Test
│
├── TWO GROUPS
│   ├── Normal + Equal Var?
│   │   ├── Independent? → Independent T-Test
│   │   └── Same subjects? → Paired T-Test
│   ├── Normal + Unequal Var? → Welch's T-Test
│   └── NOT Normal?
│       ├── Independent? → Mann-Whitney U
│       └── Same subjects? → Wilcoxon Signed-Rank
│
├── THREE+ GROUPS
│   ├── Normal + Equal Var? → One-Way ANOVA → Tukey HSD
│   ├── Two factors? → Two-Way ANOVA
│   ├── Same subjects? → Repeated Measures ANOVA
│   └── NOT Normal? → Kruskal-Wallis
│
└── RELATIONSHIP?
    ├── Both numeric + Normal? → Pearson Correlation
    ├── Numeric + NOT Normal? → Spearman Correlation
    ├── Both categorical? → Chi-Square Test
    └── Prediction?
        ├── Predict number? → Linear Regression
        └── Predict category? → Logistic Regression


━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
NEW ADDITIONS
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

20. Z-TEST
    ├── WHEN : Large sample (n > 30) + population std dev KNOWN
    ├── WHY  : Like t-test but for large samples with known spread
    ├── HOW  : p < 0.05 → Sample mean IS different from population
    ├── DIFF : T-test = small sample, unknown std
    │          Z-test = large sample, known std
    └── CODE :
             from statsmodels.stats.weightstats import ztest
             stat, p = ztest(data, value=population_mean)
             # Example: Factory produces bolts, mean length should be
             # 50mm. Test 100 bolts to verify.

21. IQR (Interquartile Range) - OUTLIER DETECTION
    ├── WHEN : Finding and removing outliers in any dataset
    ├── WHY  : Outliers can ruin statistical tests, IQR finds them
    ├── HOW  : Anything below Q1-1.5*IQR or above Q3+1.5*IQR = Outlier
    └── CODE :
             Q1  = np.percentile(data, 25)   # 25th percentile
             Q3  = np.percentile(data, 75)   # 75th percentile
             IQR = Q3 - Q1                   # Middle 50% spread
             lower_fence = Q1 - 1.5 * IQR   # Below this = outlier
             upper_fence = Q3 + 1.5 * IQR   # Above this = outlier
             clean_data  = data[(data >= lower_fence) &
                                (data <= upper_fence)]
             # Example: Remove extreme salary values before analysis
"""

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
GOLDEN RULES TO REMEMBER
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

✅ p < 0.05 → Result is statistically significant (not by chance)
✅ p > 0.05 → Result is NOT significant (could be random)
✅ Always check normality BEFORE choosing a test
✅ Use non-parametric tests when data is skewed/not normal
✅ ANOVA tells you IF difference exists → Post-hoc tells WHERE
✅ Correlation ≠ Causation (just because related doesn't mean one causes other)
✅ Larger sample size → More reliable results
"""
```
