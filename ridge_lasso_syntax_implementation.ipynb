```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(45)
```

```python
n = 4000
X1 =np.random.normal(50,10,n)
X2 = X1 * 0.85 + np.random.normal(0, 3, n)
X3 = np.random.normal(30, 8, n)
X4 = X3 * 0.75 + np.random.normal(0, 4, n)
X5 = np.random.normal(100, 20, n)
X6 = np.random.normal(0, 1, n)
X7 = np.random.normal(0, 1, n)
X8 = np.random.normal(0, 1, n)
X9 = np.random.normal(0, 1, n)
X10 = np.random.normal(0, 1, n)

############# EXPLANATION -> WE CREATED X1 AS AN ARRAY WHERE MEAN IS 50 STD IS 10 AND NO OF ELEMENTS IS N WHICH IS 4000
############################ THEN WE DEFINED FEATURES SOME OF WHICH  ARE CORELATED TO X1 AND SOME ARE NOT
############################# NOW THIS WORKS AS OUR DATASET IN THIS LEARNING FILE
```

```python
######################### NOW WE WILL CREATE A TARGET COLUMN FOR OUR DATASET
noise = np.random.normal(0, 15, n)
y = (5* X1 +3*X2+8*X3+4*X4+2*X5+noise)
###### here we created a target column for our dataset we will predict which depends on X1,X2,X3,X4,X5 AND noise
###### we  CREATED noise because if we didn't then out output would be 100% dependent on these features which is
######  doesnot hapeen in real life, there is always uncertainity,so it cant be 100%dependent on the features.
df = pd.DataFrame({ "X1": X1,"X2": X2, "X3": X3, "X4": X4,  "X5": X5,"X6": X6,"X7": X7,"X8": X8, "X9": X9, "X10": X10,"y": y})
```

```python
df.info()
```

```python
df.corr()
```

```python
# now we will separate the input and target columns
x =df.drop("y", axis=1)
y =df["y"]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.3,random_state= 45)
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_scaled =  scaler.fit_transform(x_train)
x_test_scaled =  scaler.transform(x_test)
```

```python
print(x_train_scaled)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
```

```python
lr = LinearRegression()
lr.fit(x_train_scaled, y_train)
```

```python
print(lr.coef_)
```

```python
print(lr.intercept_)
```

```python
y_pred_lr = lr.predict(x_test_scaled)
```

```python
print(y_pred_lr)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
mse_lr = mean_squared_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mse_lr)
r2_lr = r2_score(y_test, y_pred_lr)
print("MSE:", mse_lr)
print("RMSE:", rmse_lr)
print("R²:", r2_lr)
```

```python
ridge = Ridge(alpha=1.0)
ridge.fit(x_train_scaled, y_train)
y_pred_ridge = ridge.predict(x_test_scaled)
```

```python
y_pred_ridge
```

```python
print(ridge.coef_)

###### AND HERE YOU CAN SEE THAT RIDGE CAN ONLY SHRINK THE NON CORELATED FEATURES BUT CANNOT TURN THEM TO 0
```

```python
print("Linear:", lr.coef_)
print("Ridge :", ridge.coef_)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
mse_ridge = mean_squared_error(y_test, y_pred_ridge)
rmse_ridge = np.sqrt(mse_ridge)
r2_ridge = r2_score(y_test, y_pred_ridge)
print("MSE:", mse_ridge)
print("RMSE:", rmse_ridge)
print("R²:", r2_ridge)
```

```python
print("Linear Regression")
print("RMSE:", rmse_lr)
print("R²:", r2_lr)
print("\nRidge")
print("RMSE:", rmse_ridge)
print("R²:", r2_ridge)
```

```python
############### now we will try a bug alpha value to see the ridge's proper effect
ridge_strong = Ridge(alpha=100)
ridge_strong.fit(x_train_scaled, y_train)
y_pred_ridge_2 = ridge_strong.predict(x_test_scaled)
```

```python
print(ridge_strong.coef_)
```

```python
from sklearn.linear_model import Lasso
lasso = Lasso(alpha=1.0)
lasso.fit(x_train_scaled, y_train)
y_pred_lasso = lasso.predict(x_test_scaled)
```

```python
print(lasso.coef_)
####### HERE ONE THING YOU WILL NOTICE THAT LASSO CAN POTENTIALLY  SHRINK DOWNTHE LESS OR VERY WEAKLY
####### CORRELATED FEATURES TO ZERO, WHICH HELPS US IN THE FEATURE SELECTION
```

```python
print("Linear :", lr.coef_)
print("Ridge  :", ridge.coef_)
print("Lasso  :", lasso.coef_)
# here you can  SEE THE  difference between them properly
```
