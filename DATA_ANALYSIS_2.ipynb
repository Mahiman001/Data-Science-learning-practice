```python
## now we will fix the missing value and null values in the age column  ##
```

```python
import re
def extract_age(age):
 age_num = re.findall('[0-9]+', str(age))
 if len(age_num)>0:
  return age_num[0]
 else:
    return age

df['Age'] = df['Age'].apply(lambda x:extract_age(x))
```

```python
df['Age'].unique()
```

```python
df_age= df[df['Age'] != 'nan years']['Age']
```

```python
age_median = df_age.dropna().astype('int64').median()
```

```python
df.replace('nan years',age_median, inplace = True )
```

```python
df['Age'] = df['Age'].fillna(age_median)
```

```python
df['Age'].unique()
```

```python
#   df['Purchase_Amount'].fillna(df['Purchase_Amount'].median(), inplace =True )


## THIS METHOD WORKS BUT NOT GOOD BCUZ
# inplace=True on df['column'] is discouraged because df['column'] may return a copy instead of the original data,
##   so pandas cannot guarantee that the DataFrame will actually be modified.
##  Assigning the result back to the column removes this ambiguity.
```

```python
## YOU SHOULD DO THIS INSTEAD

df['Purchase_Amount'] = df['Purchase_Amount'].fillna(df['Purchase_Amount'].median())
```
